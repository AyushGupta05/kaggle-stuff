```python
from sklearn.datasets import fetch_openml
from sklearn.preprocessing import OrdinalEncoder
```

```python
data = fetch_openml("car",as_frame = True).frame
```

```python
data
```

```python
columns_to_encode = ["lug_boot", "safety"]
encoder = OrdinalEncoder(
    categories = [
        ["small", "med", "big"],
        ["low", "med", "high"]
    ]
)

data[columns_to_encode] = encoder.fit_transform(data[columns_to_encode])

```

```python
data
```

```python
encoder.inverse_transform(data[columns_to_encode])

# stores them back to what they were initially 
```

```python
# one hot encoder 

data2 = fetch_openml("adult",as_frame=True).frame
```

```python
data2.head(6)
```

```python
from sklearn.preprocessing import OneHotEncoder

data2["occupation"].value_counts()
```

```python
import pandas as pd 

encoder = OneHotEncoder(handle_unknown="ignore", sparse_output = False)

encoded_values = encoder.fit_transform(data2[["occupation","race"]])
new_cols = encoder.get_feature_names_out(["occupation","race"])
```

```python
df_encoded = pd.DataFrame(encoded_values,columns = new_cols, index = data2.index)
data2 = pd.concat(
    [data2.drop(columns = ["occupation", "race"]), df_encoded], axis = 1
    )
```
