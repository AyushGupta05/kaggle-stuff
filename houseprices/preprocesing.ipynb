```python
import pandas as pd 

data = pd.read_csv("data/train.csv")
test= pd.read_csv("data/test.csv")
test_ids = test["Id"]

test_ids.to_csv("test_ids.csv", index=False)
```

```python
df = [data,test]

for value in df:
    value.drop(["Id","Alley"], axis = 1, inplace = True)
    value["MSSubClass"] = value["MSSubClass"].astype(str)
```

```python
neighborhoods = {}       


for neighborhood in data["Neighborhood"].unique():
    neighborhoods[neighborhood] = data[data["Neighborhood"] == neighborhood]

```

```python
zero_cols = [
    "MasVnrArea",
    "BsmtFullBath",
    "BsmtHalfBath",
    "GarageCars",
    "GarageArea"
]

for value in df:
    value[zero_cols] = value[zero_cols].fillna(0)

```

```python
num_cols = data.select_dtypes(include="number").columns.drop("SalePrice")
neighborhood_medians = (data.groupby("Neighborhood")[num_cols].median())

for col in num_cols:
    data[col] = data[col].fillna(
        data["Neighborhood"].map(neighborhood_medians[col])
    )
    test[col] = test[col].fillna(
        test["Neighborhood"].map(neighborhood_medians[col])
    )
```

```python
utilities_map = {
    "AllPub": 3,
    "NoSewr": 2,
    "NoSeWa": 1,
    "ELO": 0
}

landslope_map = {
    "Gtl": 3,
    "Mod": 2,
    "Sev": 1
}

exterqual_map = {
    "Ex": 4,
    "Gd": 3,
    "TA": 2,
    "Fa": 1,
    "Po": 0
}

extercond_map = {
    "Ex": 4,
    "Gd": 3,
    "TA": 2,
    "Fa": 1,
    "Po": 0
}

bsmtqual_map = {
    "Ex": 100,
    "Gd": 95,
    "TA": 85,
    "Fa": 75,
    "Po": 65,
    "NA": 0
}

bsmtcond_map = {
    "Ex": 5,
    "Gd": 4,
    "TA": 3,
    "Fa": 2,
    "Po": 1,
    "NA": 0
}

bsmtexposure_map = {
    "Gd": 4,
    "Av": 3,
    "Mn": 2,
    "No": 1,
    "NA": 0
}

bsmtfintype_map = {
    "GLQ": 6,
    "ALQ": 5,
    "BLQ": 4,
    "Rec": 3,
    "LwQ": 2,
    "Unf": 1,
    "NA": 0
}

heatingqc_map = {
    "Ex": 4,
    "Gd": 3,
    "TA": 2,
    "Fa": 1,
    "Po": 0
}

centralair_map = {
    "Y": 1,
    "N": 0
}

kitchenqual_map = {
    "Ex": 4,
    "Gd": 3,
    "TA": 2,
    "Fa": 1,
    "Po": 0
}

functional_map = {
    "Typ": 7,
    "Min1": 6,
    "Min2": 5,
    "Mod": 4,
    "Maj1": 3,
    "Maj2": 2,
    "Sev": 1,
    "Sal": 0
}

fireplacequ_map = {
    "Ex": 5,
    "Gd": 4,
    "TA": 3,
    "Fa": 2,
    "Po": 1,
    "NA": 0
}

garagefinish_map = {
    "Fin": 3,
    "RFn": 2,
    "Unf": 1,
    "NA": 0
}

garagequal_map = {
    "Ex": 5,
    "Gd": 4,
    "TA": 3,
    "Fa": 2,
    "Po": 1,
    "NA": 0
}

garagecond_map = {
    "Ex": 5,
    "Gd": 4,
    "TA": 3,
    "Fa": 2,
    "Po": 1,
    "NA": 0
}

paveddrive_map = {
    "Y": 2,
    "P": 1,
    "N": 0
}

poolqc_map = {
    "Ex": 4,
    "Gd": 3,
    "TA": 2,
    "Fa": 1,
    "NA": 0
}

fence_map = {
    "GdPrv": 4,
    "MnPrv": 3,
    "GdWo": 2,
    "MnWw": 1,
    "NA": 0
}

for value in df:
    value["Utilities"] = value["Utilities"].fillna("ELO").map(utilities_map)
    value["LandSlope"] = value["LandSlope"].map(landslope_map)
    value["ExterQual"] = value["ExterQual"].map(exterqual_map)
    value["ExterCond"] = value["ExterCond"].map(extercond_map)
    value["BsmtQual"] = value["BsmtQual"].fillna("NA").map(bsmtqual_map)
    value["BsmtCond"] = value["BsmtCond"].fillna("NA").map(bsmtcond_map)
    value["BsmtExposure"] = value["BsmtExposure"].fillna("NA").map(bsmtexposure_map)
    value["BsmtFinType1"] = value["BsmtFinType1"].fillna("NA").map(bsmtfintype_map)
    value["BsmtFinType2"] = value["BsmtFinType2"].fillna("NA").map(bsmtfintype_map)
    value["HeatingQC"] = value["HeatingQC"].map(heatingqc_map)
    value["CentralAir"] = value["CentralAir"].fillna("N").map(centralair_map)
    value["KitchenQual"] = value["KitchenQual"].fillna("TA").map(kitchenqual_map)
    value["Functional"] = value["Functional"].fillna("Typ").map(functional_map)
    value["FireplaceQu"] = value["FireplaceQu"].fillna("NA").map(fireplacequ_map)
    value["GarageFinish"] = value["GarageFinish"].fillna("NA").map(garagefinish_map)
    value["GarageQual"] = value["GarageQual"].fillna("NA").map(garagequal_map)
    value["GarageCond"] = value["GarageCond"].fillna("NA").map(garagecond_map)
    value["PavedDrive"] = value["PavedDrive"].map(paveddrive_map)
    value["PoolQC"] = value["PoolQC"].fillna("NA").map(poolqc_map)
    value["Fence"] = value["Fence"].fillna("NA").map(fence_map)
    value["GarageQual"] = value["GarageQual"].fillna(0)
    value["GarageCond"] = value["GarageCond"].fillna(0)
```

```python
rail_conditions = ["RRNn", "RRAn", "RRNe", "RRAe"]

for value in df:
    value["Lotareafrontage"] = value["LotArea"] * value["LotFrontage"]
    value["Lotfrontagearea"] = value["LotArea"]/value["LotFrontage"].replace(0, 1)
    value["LotShape"] = value["LotShape"].fillna("U")
    value["LandContour"] = value["LandContour"].fillna("U")
    value["Combinedcontourshape"] = value["LotShape"].astype(str) + "_" + value["LandContour"].astype(str)
    
    value["nearrailroad"] = (value["Condition1"].isin(rail_conditions)| value["Condition2"].isin(rail_conditions)).astype(int)
    value["CombinedCondition"] = (value["Condition1"].astype(str) + "_" + value["Condition2"].astype(str))
    value.drop(["Condition1", "Condition2"], axis = 1, inplace = True)
    value["hasfinishedsecondlevel"] = value["HouseStyle"].isin(["1.5Fin", "2Story", "2.5Fin"]).astype(int)
    value["OverallQualcond"] = (2 * value["OverallQual"] * value["OverallCond"])/(value["OverallQual"] + value["OverallCond"])
    value["Ageatsale"] = value["YrSold"] - value["YearBuilt"]
    value["Yearstoremodel"] = value["YearRemodAdd"] - value["YearBuilt"]
    value.drop(["YearRemodAdd"], axis = 1, inplace = True)
    value["Exterior"] = (value["Exterior1st"].astype(str) + "_" + value["Exterior2nd"].astype(str))
    value.drop(["Exterior1st", "Exterior2nd"], axis = 1, inplace = True)
    value["hasmasonyveer"] = (value["MasVnrArea"].fillna(0) != 0).astype(int)
    value["hasbasement"] = (value["BsmtQual"] != 0).astype(int)
    value["Finishedbasement"] = value["TotalBsmtSF"] - value["BsmtUnfSF"]
    value["hassecondfloor"] = (value["2ndFlrSF"] != 0).astype(int)
    value["haslowqualfinish"] = (value["LowQualFinSF"] != 0).astype(int)
    value["TotalBathrooms"] = (value["FullBath"]+ 0.5 * value["HalfBath"]+ value["BsmtFullBath"].fillna(0)+ 0.5 * value["BsmtHalfBath"].fillna(0))
    value["hasfireplace"] = (value["Fireplaces"] != 0).astype(int)
    value["hasgarage"] = (value["GarageCars"].fillna(0) != 0).astype(int)
    value["Garageageatsale"] = (value["YrSold"] - value["GarageYrBlt"]).fillna(0)
    value["GarageQualcond"] = (2 * value["GarageQual"] * value["GarageCond"])/(value["GarageQual"] + value["GarageCond"]).replace(0, 1)
    value["haswooddeck"] = (value["WoodDeckSF"] != 0).astype(int)
    value["hasopenporch"] = (value["OpenPorchSF"] != 0).astype(int)
    value["hasenclosedporch"] = (value["EnclosedPorch"] != 0).astype(int)
    value["has3seasonporch"] = (value["3SsnPorch"] != 0).astype(int)
    value["hasscreenporch"] = (value["ScreenPorch"] != 0).astype(int)
    value["haspool"] = (value["PoolArea"] != 0).astype(int)
    value["hasfence"] = (value["Fence"] != 0).astype(int)
    value["haselevator"] = (value["MiscFeature"] == "Elev").astype(int)
    value["hassecondgarage"] = (value["MiscFeature"] == "Gar2").astype(int)
    value["hasshed"] = (value["MiscFeature"] == "Shed").astype(int)
    value["hastenniscourt"] = (value["MiscFeature"] == "TenC").astype(int)
```

```python
from sklearn.preprocessing import OneHotEncoder
word_cols = data.select_dtypes(include=["object", "string"]).columns
encoder = OneHotEncoder(handle_unknown="ignore", sparse_output=False)
encoder.fit(data[word_cols])
train_encoded = encoder.transform(data[word_cols])
test_encoded = encoder.transform(test[word_cols])
```

```python

train_encoded = pd.DataFrame(
    train_encoded,
    columns=encoder.get_feature_names_out(word_cols),
    index=data.index
)

test_encoded = pd.DataFrame(
    test_encoded,
    columns=encoder.get_feature_names_out(word_cols),
    index=test.index
)

data = pd.concat(
    [data.drop(word_cols, axis=1), train_encoded],
    axis=1
)

test = pd.concat(
    [test.drop(word_cols, axis=1), test_encoded],
    axis=1
)
```

```python
import numpy as np

for value in [data, test]:
    print("Non numeric:")
    print(value.select_dtypes(exclude="number").columns)

    print("Missing:")
    missing_values = value.isna().sum()
    print(missing_values[missing_values > 0])

    print("Infinite:")
    print(np.isinf(value.select_dtypes(include="number")).sum().sum())
```
